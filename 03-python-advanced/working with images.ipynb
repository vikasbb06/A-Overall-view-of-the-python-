Color transparency

RGBA-RED,GREEN,BLUE,ALPHA

```python
red=Image.open("red_color.jpg")
red
```

```python
red.putalpha(100) # 0 is the complete transparency of the image & large number (300) is the complete opaque of the image
red
```

```python
blue=Image.open("blue_color.png")
blue
```

```python
# Convert the image so the notebook can display it
blue = blue.convert("RGBA")
blue.putalpha(100)
blue
```

```python
blue.paste(im=red,box=(0,0),mask=red)
blue
```
