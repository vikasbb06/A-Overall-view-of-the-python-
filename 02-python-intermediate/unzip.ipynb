```python
import os
import re

pattern = r'\d{3}-\d{3}-\d{4}'

def search(file_path,pattern):
    # 'with' automatically closes the file even if an error happens
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            text = f.read()
            match = re.search(pattern, text)
            return match if match else ""
    except (UnicodeDecodeError, PermissionError):
        # Safely skip binary, hidden, or system files
        return ""

results = []
target_dir = os.path.join(os.getcwd(), 'unzip')

# Walk through all directories
for root, sub_folders, files in os.walk(target_dir):
    for file in files:
        full_path = os.path.join(root, file)
        match_obj = search(full_path, pattern)
        if match_obj: # Only store actual match objects
            results.append(match_obj)

# Out of the loop: Print everything once at the very end
for r in results:
    if r!="":
        print(r.group())
```
